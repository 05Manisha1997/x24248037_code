# 05 — Multimodal Fusion (Text + Image + Video)


```python
import sys
from pathlib import Path
ROOT = Path('..').resolve()
sys.path.insert(0, str(ROOT))

import subprocess
import pandas as pd
from src.utils import ensure_dir, load_config

config = load_config(ROOT / "config.yaml")
results_dir = ensure_dir(ROOT / "outputs" / "results")
```

```python
# Trimodal LMVD experiments (text + image + video)
print("Running trimodal experiments on LMVD (text + image + video)...")
subprocess.run(
    [sys.executable, str(ROOT / "scripts" / "run_trimodal_experiments.py"), "--fast",
     "--skip-text", "--skip-image", "--skip-video", "--skip-bimodal"],
    cwd=str(ROOT),
)

trimodal_path = results_dir / "trimodal_results.csv"
if trimodal_path.exists():
    trimodal_df = pd.read_csv(trimodal_path)
    print(f"\nTrimodal results ({len(trimodal_df)} models):")
    cols = [c for c in ("modality", "model", "test_accuracy", "test_f1_macro", "test_f1_weighted") if c in trimodal_df.columns]
    print(trimodal_df.sort_values("test_accuracy", ascending=False).head(10)[cols].to_string(index=False))
else:
    print("Trimodal results not found — ensure LMVD features are downloaded.")
```

```python
# Text + image-proxy fusion via main experiment runner (fair metrics)
print("Running proxy fusion (early/late/attention) with fair text baseline...")
subprocess.run(
    [
        sys.executable, str(ROOT / "scripts" / "run_experiments.py"),
        "--modalities", "fusion", "--text-sample", "3000", "--quick",
    ],
    cwd=str(ROOT),
)
fusion_path = results_dir / "fusion_results.csv"
if fusion_path.exists():
    print(pd.read_csv(fusion_path).to_string(index=False))
print("05_multimodal_fusion.py complete.")
```
