# 01 — Data Setup & Downloads


```python
import sys
from pathlib import Path

ROOT = Path('..').resolve()
sys.path.insert(0, str(ROOT))

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
print("Project root:", ROOT)
```

```python
import subprocess
result = subprocess.run(
    [sys.executable, str(ROOT / "scripts" / "download_all.py"), "--skip-lmvd"],
    cwd=str(ROOT), capture_output=True, text=True,
)
print(result.stdout[-800:] if len(result.stdout) > 800 else result.stdout)
```

```python
from IPython.display import display
from src.data.loaders import load_structured_data, load_text_data, load_lmvd_labels

structured = load_structured_data()
text_df = load_text_data(sample=5000)
lmvd_labels = load_lmvd_labels(ROOT / "lmvd-analysis" / "data")

print("Structured:", structured.shape)
print("Text sample:", text_df.shape)
print("LMVD labels:", lmvd_labels.shape)
display(structured.head(3))
display(text_df[["text", "status", "aosp_composite", "wellbeing_decline"]].head(3))
```

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4))
structured["GAD_7_Score"].hist(ax=axes[0], bins=20, color="#4C72B0")
axes[0].set_title("GAD-7 Distribution")
structured["PHQ_9_Score"].hist(ax=axes[1], bins=20, color="#DD8452")
axes[1].set_title("PHQ-9 Distribution")
lmvd_labels["class"].value_counts().plot(kind="bar", ax=axes[2], color=["#4C72B0", "#DD8452"])
axes[2].set_title("LMVD Class Balance")
plt.tight_layout()
out = ROOT / "outputs"
out.mkdir(exist_ok=True)
plt.savefig(out / "data_overview.png", dpi=150, bbox_inches="tight")
plt.close()
print("Saved outputs/data_overview.png")
```
